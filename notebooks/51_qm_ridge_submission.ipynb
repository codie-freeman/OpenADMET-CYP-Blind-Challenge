# 51 -- free 2D plus the quantum descriptors, ridge, one submission candidate

One of a **matched pair**. Notebook 50 fits the 226 free 2D descriptors; notebook 51 fits those
same 226 plus the 13 molecule-level QM columns and notebook 45's 4 atom descriptors. They are
sent about 12 hours apart and **the feature block is the only thing that differs between them**.
If anything else differed, a board difference would not be attributable to the quantum data and
the exercise would be wasted.

**How the pairing is guaranteed, structurally rather than by inspection.** There is exactly one
implementation of the fit, the correction, the validation and the output format. It lives in
`scripts/50_ridge_submission.py` and both notebooks call `build(block=...)` with nothing else
changed. That script writes a `pipeline_fingerprint.json` recording every parameter it used, and
**Part 9 of this notebook loads notebook 50's fingerprint and asserts equality on every field in `MUST_BE_IDENTICAL`.** Notebook 50 must have been run first.

**What is shared, explicitly:**

| | |
|---|---|
| estimator | notebook 46's own `make_model("ridge", ...)`, imported not retyped: `Pipeline[clip to the fit rows' 1st/99th percentile, StandardScaler, RidgeCV]` |
| alpha grid | notebook 46's `RIDGE_ALPHAS`, `logspace(-2, 4, 13)` |
| alpha selection | RidgeCV's internal leave-one-out, per endpoint, on training rows only |
| seed | 2684470948, recorded for the fingerprint; ridge is deterministic at these settings and the fit is checked twice under different seeds to prove it |
| training rows | all 4,905 curated compounds; per endpoint, the rows carrying a label for it |
| blind rows | the 750, from `outputs/49_qm_only_arm/features_blind_extended/`, built by notebook 49 and asserted there to leave training values unchanged |
| correction | two layers, identical procedure: CYP2D6 placement onto notebook 16's solved blind population using **this model's own** out-of-fold rho, then notebook 36's widening to the best-board-column ratio where the relative gap reaches 10% |
| output | `SMILES`, `Molecule_Name` and the four endpoints; validated by the unmodified `validate_activity_submission` |

**What legitimately differs:** the feature block, and the correction *parameters* (alpha, the
out-of-fold rho, the widening multiplier), which are fitted per model. That is what a fitted
correction is for. The *procedure* producing them is the same code.

**One ridge per endpoint**, which is forced by the labelling: the four endpoints are 26-48%
sparse and no single-output ridge takes a NaN target. This is not per-isoform model selection.
The block, the estimator, the alpha grid and the clip are fixed and identical across endpoints
and across the two notebooks. No blending anywhere.

`SEND = False`. Nothing is sent. `notebooks/32_deadzone_aid_retrain.ipynb` and
`notebooks/40_single_object_ensemble.ipynb` are never opened.

```python
# Part 0 -- setup, provenance, and the environment assertions this project requires.
import hashlib, importlib.util, json, os, subprocess, sys, warnings
from datetime import datetime, timezone
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")
REPO = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(REPO))

assert os.environ.get("OMP_NUM_THREADS") is None, "OMP_NUM_THREADS must be unset"
print("OMP_NUM_THREADS      : unset (asserted)")
print(f"KMP_DUPLICATE_LIB_OK : {os.environ.get('KMP_DUPLICATE_LIB_OK')!r} (reported, not asserted)")

BLOCK = "FREE_QM13_ATOM4"
ARM = "free_qm13_atom4__ridge"
OUT = REPO / "outputs" / "51_qm_ridge_submission"
PAIR_OUT = REPO / "outputs" / "50_free_ridge_submission"
MODEL_TAG = ("Ridge on the 226 free 2D descriptors plus 13 molecule-level QM columns (B3LYP D3 "
             "def2-SV(P) Hirshfeld, via GFN2 geometries) and 4 atom-level QM descriptors, "
             "one fit per endpoint on all 4,905 training compounds, RidgeCV alpha. CYP2D6 "
             "placement-corrected onto the board-solved blind population using this model's own "
             "out-of-fold rho; CYP1A2/CYP2D6/CYP3A4 spread-widened to each isoform's "
             "best-board-column ratio; CYP2C9 uncorrected. Matched pair with the free-only "
             "variant, feature block the only difference.")
FIG = OUT / "figures"; FIG.mkdir(parents=True, exist_ok=True)
ISOFORMS = ["CYP1A2", "CYP2C9", "CYP2D6", "CYP3A4"]
EP = {i: f"{i}_pIC50_direct_inhibition" for i in ISOFORMS}

def sha256(p):
    h = hashlib.sha256()
    with open(p, "rb") as f:
        for c in iter(lambda: f.read(1 << 20), b""):
            h.update(c)
    return h.hexdigest()

RUN_STARTED_AT = datetime.now(timezone.utc)
_pre = subprocess.run(["git", "status", "--porcelain"], cwd=REPO, capture_output=True, text=True)
PRE_DIRTY = sorted({l[3:].strip() for l in _pre.stdout.splitlines() if not l.startswith("??")})
print(f"\nrun started {RUN_STARTED_AT.isoformat()}")
print(f"tracked paths already dirty before this run: {len(PRE_DIRTY)}")
for p in PRE_DIRTY:
    print("   ", p)
print(f"\nfeature block under test: {BLOCK}   reference arm: {ARM}")
```

```python
# Part 1 -- the shared implementation, imported.
SPEC = REPO / "scripts" / "50_ridge_submission.py"
spec = importlib.util.spec_from_file_location("ridge_submission", SPEC)
SUB = importlib.util.module_from_spec(spec); spec.loader.exec_module(SUB)
print(f"implementation : {SPEC.relative_to(REPO)}")
print(f"  sha256       : {sha256(SPEC)}")
print(f"notebook 46 runner it imports for make_model:")
print(f"  sha256       : {sha256(REPO / 'scripts' / '46_run_arms.py')}")
m46 = SUB.nb46_module()
print(f"\nridge alphas  : {[round(float(a), 4) for a in m46.RIDGE_ALPHAS]}")
print(f"clip percentile: {m46.RIDGE_CLIP_PCT}")
print(f"seed          : {SUB.SEED}  (ridge is deterministic at these settings; checked in the fit)")
print(f"widening bar  : {SUB.WIDEN_BAR}")
print(f"CYP2D6 target : mean {SUB.NB16_MU}  sd {SUB.NB16_SD}  (notebook 16's deployed solve)")
```

## Part 9 -- the assertion that this pipeline matches notebook 50's

This is the check the whole exercise rests on. Notebook 50's `pipeline_fingerprint.json` records
every parameter its run used; this cell loads it and compares field by field against this
notebook's. Anything in `MUST_BE_IDENTICAL` that differs fails the assertion.

It also checks the consequence that makes the pair read cleanly: after correction, the two files
are forced to the **same** spread on CYP1A2, CYP2D6 and CYP3A4, and to the same **mean** as well
on CYP2D6. On those three isoforms the two submissions can differ only in rank order.

```python
# Part 9 -- assert this pipeline is notebook 50's, except for the feature block.
P50 = PAIR_OUT / "pipeline_fingerprint.json"
assert P50.exists(), (
    f"{P50.relative_to(REPO)} is missing. Notebook 50 must be run before notebook 51, or the "
    f"pairing cannot be verified.")
A = json.loads(P50.read_text())
B = FP

print("top-level implementation identity:")
for k in ("implementation", "implementation_sha256", "nb46_runner_sha256"):
    same = A[k] == B[k]
    print(f"  {k:24s} {'MATCH' if same else 'DIFFERS'}   {str(B[k])[:64]}")
    assert same, k

print("\nMUST_BE_IDENTICAL, field by field:")
assert set(A["MUST_BE_IDENTICAL"]) == set(B["MUST_BE_IDENTICAL"]), "field sets differ"
for k in sorted(A["MUST_BE_IDENTICAL"]):
    a, b = A["MUST_BE_IDENTICAL"][k], B["MUST_BE_IDENTICAL"][k]
    same = a == b
    shown = str(b) if len(str(b)) <= 70 else str(b)[:67] + "..."
    print(f"  {k:28s} {'MATCH' if same else 'DIFFERS'}   {shown}")
    assert same, f"{k} differs: {a!r} vs {b!r}"

print("\nVARIES_BETWEEN_NOTEBOOKS, which is the whole point:")
for k in sorted(A["VARIES_BETWEEN_NOTEBOOKS"]):
    print(f"  {k:14s} nb50 {str(A['VARIES_BETWEEN_NOTEBOOKS'][k]):44s} "
          f"nb51 {B['VARIES_BETWEEN_NOTEBOOKS'][k]}")
assert A["VARIES_BETWEEN_NOTEBOOKS"]["block"] != B["VARIES_BETWEEN_NOTEBOOKS"]["block"]

print("\nFITTED_PER_MODEL, legitimately different because it is fitted:")
for k in sorted(A["FITTED_PER_MODEL"]):
    print(f"  {k}")
    print(f"    nb50 {A['FITTED_PER_MODEL'][k]}")
    print(f"    nb51 {B['FITTED_PER_MODEL'][k]}")

print("\nTHE PIPELINES ARE IDENTICAL APART FROM THE FEATURE BLOCK (asserted on every field).")
```

```python
# Part 9.2 -- what the correction leaves comparable, and what it isolates.
a_raw = pd.read_csv(PAIR_OUT / "submission_candidate_raw.csv")
a_cor = pd.read_csv(PAIR_OUT / "submission_candidate.csv")
b_raw = pd.read_csv(OUT / "submission_candidate_raw.csv")
b_cor = pd.read_csv(OUT / "submission_candidate.csv")
assert (a_cor.Molecule_Name == b_cor.Molecule_Name).all()
rows = []
for iso in ISOFORMS:
    ar, br = a_cor[EP[iso]].to_numpy(), b_cor[EP[iso]].to_numpy()
    rows.append(dict(isoform=iso,
                     nb50_mean=ar.mean(), nb51_mean=br.mean(), mean_diff=br.mean() - ar.mean(),
                     nb50_sd=ar.std(ddof=1), nb51_sd=br.std(ddof=1),
                     sd_diff=br.std(ddof=1) - ar.std(ddof=1),
                     pearson=float(np.corrcoef(ar, br)[0, 1]),
                     spearman=float(pd.Series(ar).corr(pd.Series(br), method="spearman")),
                     mean_abs_diff=float(np.abs(ar - br).mean()),
                     n_moving_gt_0p5=int((np.abs(ar - br) > 0.5).sum())))
CMP = pd.DataFrame(rows); CMP.to_csv(OUT / "pair_comparison.csv", index=False)
print("the two CORRECTED submissions, compared directly:\n")
print(CMP.round(5).to_string(index=False))
forced = [r.isoform for _, r in CMP.iterrows() if abs(r.sd_diff) < 1e-9]
print(f"\nisoforms where the correction forced both files to the SAME spread: {forced}")
same_mean = [r.isoform for _, r in CMP.iterrows() if abs(r.mean_diff) < 1e-9]
print(f"isoforms where it also forced the SAME mean: {same_mean}")
print("\nOn those, the two submissions differ only in rank order, which is the quantity notebook")
print("38 identified as what separates this project's leading columns and which no affine")
print("correction can reach. Spearman between the two corrected files is "
      + ", ".join(f"{r.isoform} {r.spearman:.4f}" for _, r in CMP.iterrows()) + ".")
print("\nThe raw files, for completeness (no correction, so spread differs freely):")
rr = pd.DataFrame([dict(isoform=i,
                        sd_diff=float(b_raw[EP[i]].std(ddof=1) - a_raw[EP[i]].std(ddof=1)),
                        mean_abs_diff=float(np.abs(b_raw[EP[i]] - a_raw[EP[i]]).mean()),
                        spearman=float(a_raw[EP[i]].corr(b_raw[EP[i]], method="spearman")))
                   for i in ISOFORMS])
print(rr.round(5).to_string(index=False))
```
